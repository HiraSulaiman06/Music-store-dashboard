## Prepare Data for streamlit:

```python
import pandas as pd

customer = pd.read_csv("customer.csv")
invoice = pd.read_csv("invoice.csv")
invoice_line = pd.read_csv("invoice_line.csv")
track = pd.read_csv("track.csv")
genre = pd.read_csv("genre.csv")
```

```python
# Clean the important columns:
```

```python
invoice["invoice_date"] = pd.to_datetime(invoice["invoice_date"])

invoice_line["unit_price"] = pd.to_numeric(invoice_line["unit_price"], errors="coerce")
invoice_line["quantity"] = pd.to_numeric(invoice_line["quantity"], errors="coerce")

invoice_line["revenue"] = (
    invoice_line["unit_price"] * invoice_line["quantity"]
)

genre = genre.rename(columns={"name": "genre_name"})
```

```python
# Combine the tables:
```

```python
music_df = (
    invoice
    .merge(invoice_line, on="invoice_id", how="inner")
    .merge(track, on="track_id", how="inner", suffixes=("", "_track"))
    .merge(genre, on="genre_id", how="left")
    .merge(
        customer[["customer_id", "first_name", "last_name", "email"]],
        on="customer_id",
        how="left"
    )
)
```

```python
# Keep only useful columns:
```

```python
music_df = music_df[
    [
        "invoice_id",
        "invoice_date",
        "customer_id",
        "first_name",
        "last_name",
        "billing_country",
        "billing_city",
        "track_id",
        "name",
        "genre_name",
        "unit_price",
        "quantity",
        "revenue"
    ]
]
```

```python
music_df = music_df.rename(columns={"name": "track_name"})
```

```python
print(music_df.shape)
display(music_df.head())
print(music_df.isnull().sum())
```

```python
music_df.to_csv("music_combined.csv", index=False)

print("✅ music_combined.csv created successfully!")
```
